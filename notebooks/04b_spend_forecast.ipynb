```python
"""04b: Spend forecast, monthly spend by department."""

from pathlib import Path

import pandas as pd

from expense_agent.paths import GENERATED_DIR

HISTORY_FILE = GENERATED_DIR / "spend_history.csv"
DEPARTMENTS = ["Engineering", "Finance", "Marketing", "Operations", "Sales"]
EXPECTED_WEEKS = 156 # 01 section 7a: 156 weeks x 5 depts (Session 4)

def load_weekly_spend(path: Path) -> pd.DataFrame:
    """Read the weekly spend history as on column per department.

    Args:
        path: CSV with columns week_start, department, spend (long format).

    Returns:
        DataFrame indexed by week_start (a Monday), one column per
        department, spend in dollars.
    """
    long_df = pd.read_csv(path, parse_dates=["week_start"])
    return long_df.pivot(index="week_start", columns="department", values="spend")

weekly = load_weekly_spend(HISTORY_FILE)


# --- checks ---
every_monday = pd.date_range(weekly.index.min(), weekly.index.max(), freq="W-MON")
assert weekly.index.equals(every_monday), "gap or non-Monday in the weekly index"
assert len(weekly) == EXPECTED_WEEKS
assert list(weekly.columns) == DEPARTMENTS
assert weekly.notna().all().all(), "zero or negative spend"

print(f"{len(weekly)} weeks: {weekly.index.min():%Y-%m-%d} -> {weekly.index.max():%Y-%m-%d}")
print(f"Total spend: ${weekly.to_numpy().sum():,.2f}")
weekly.head()
```
